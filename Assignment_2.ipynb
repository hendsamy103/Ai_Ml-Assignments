1.Write a function that takes a list of strings and a number K. The function should
return a new list containing only the strings that have a length greater than K.
 Input: list=["apple", "pie", "banana"], K=3
 Output: ['apple', 'banana']

```python

words = ["apple", "pie", "banana"]
K = 3

def check(words, k):
    new_list = []

    for x in words:  
        
        if len(x) > k:
            new_list.append(x)

    return new_list


print(check(words, K))
```

2. Write a function named area() that calculates the area of either a "circle" or a
"rectangle". The function should take the shape name as the first argument,
followed by the necessary dimensions (radius for circle; length and width for
rectangle).
 Example Input: area("circle", 5) OR area("rectangle", 4, 6)
 Example Output: Circle Area: 78.5, Rectangle Area: 24

```python
def area(shape, x, y=0):
    if shape == "circle":
        return 3.14 * x * x

    elif shape == "rectangle":
        return x * y


circle = area("circle", 5)
rectangle = area("rectangle", 4, 6)

print("Circle Area:", circle, ", Rectangle Area:", rectangle)
```

5. Write a Python class Restaurant with attributes like menu_items, book_table, and
customer_orders, and methods like add_item_to_menu, book_tables, and
customer_order. (Note: Use dictionaries and lists to store the data.)
Perform the following tasks now:
 Now add items to the menu.
 Make table reservations.
 Take customer orders.
 Print the menu.
 Print table reservations.
 Print customer orders

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    # Add items to the menu
    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    # Book a table
    def book_tables(self, customer_name, table_number):
        reservation = {
            "customer": customer_name,
            "table": table_number
        }
        self.book_table.append(reservation)

    # Take customer orders
    def customer_order(self, customer_name, items):
        order = {
            "customer": customer_name,
            "items": items
        }
        self.customer_orders.append(order)

    # Print menu
    def print_menu(self):
        print("Restaurant Menu:")
        for item, price in self.menu_items.items():
            print(item, ":", price)

    # Print reservations
    def print_reservations(self):
        print("Table Reservations:")
        for reservation in self.book_table:
            print(reservation)

    # Print customer orders
    def print_orders(self):
        print("Customer Orders:")
        for order in self.customer_orders:
            print(order)


# Create restaurant object
restaurant = Restaurant()

# Add menu items
restaurant.add_item_to_menu("Pizza", 150)
restaurant.add_item_to_menu("Burger", 100)
restaurant.add_item_to_menu("Pasta", 120)

# Make table reservations
restaurant.book_tables("Ahmed", 1)
restaurant.book_tables("Sara", 2)

# Take customer orders
restaurant.customer_order("Ahmed", ["Pizza", "Pasta"])
restaurant.customer_order("Sara", ["Burger"])

# Print all information
restaurant.print_menu()
restaurant.print_reservations()
restaurant.print_orders()
```

6. A Bike Rental System Let us start writing the class Customer that is composed of the following
attributes:
 name
 age
Here are some of the methods that you need to add to your class Customer;
 def __init__(self).
 Getter / setter for all attributes.
class MainBikeRental that is composed of the following attributes:
 List of bikes
 List of Rentals
Here are some of the methods that you need to add to your class MainBikeRental;
 def __init__(self).
 Getter / setter for all attributes.
 requestBike: Takes a request from the customer for the number of bikes and check if
the requested number is available or not if yes, function will deduct number_of_bikes
from shop stock else it will print wait 10 mins. And you have to make sure that the age
of the customer is over than 6 years.
 returnBike: Allows customers to return their bikes to the rental shop by taking
rentalTime, num_of_bikes as parameters. And calculate the total bill cost by multiplying
rentalTime * 40. Also, add the returned bikes to the stock store again
 addBike: adds bikes into the list
 totalCost: returns the total cost of bikes in the store
class Bike that is composed of the following attributes:
 name
 price
Here are some of the methods that you need to add to your class Bike;
 def __init__(self).
 Getter / setter for all attributes

```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age


class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price


class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    # Add a bike to the shop
    def addBike(self, bike):
        self.bikes.append(bike)

    # Request bikes
    def requestBike(self, customer, number_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6")
            return

        if number_of_bikes <= 0:
            print("Invalid number of bikes")
            return

        if number_of_bikes <= len(self.bikes):
            rented_bikes = []

            for i in range(number_of_bikes):
                bike = self.bikes.pop()
                rented_bikes.append(bike)

            self.rentals.append({
                "customer": customer.get_name(),
                "bikes": rented_bikes
            })

            print("Bikes rented successfully")
        else:
            print("Wait 10 mins")

    # Return bikes
    def returnBike(self, rentalTime, num_of_bikes):
        if rentalTime < 0 or num_of_bikes <= 0:
            print("Invalid return details")
            return

        for rental in self.rentals:
            if len(rental["bikes"]) == num_of_bikes:
                self.bikes.extend(rental["bikes"])
                self.rentals.remove(rental)

                total_bill = rentalTime * 40
                print("Total bill:", total_bill)
                return total_bill

        print("Rental not found")

    # Calculate total stock value
    def totalCost(self):
        total = 0
        for bike in self.bikes:
            total += bike.get_price()
        return total


# Create rental shop
shop = MainBikeRental()

# Create bikes
bike1 = Bike("Mountain Bike", 1000)
bike2 = Bike("Road Bike", 1500)
bike3 = Bike("City Bike", 800)

# Add bikes to shop
shop.addBike(bike1)
shop.addBike(bike2)
shop.addBike(bike3)

# Create customer
customer1 = Customer("Ahmed", 20)

# Print total stock value
print("Total stock value:", shop.totalCost())

# Rent 2 bikes
shop.requestBike(customer1, 2)

# Check available bikes
print("Available bikes:", len(shop.get_bikes()))

# Return 2 bikes after 3 hours
shop.returnBike(3, 2)

# Check stock again
print("Available bikes:", len(shop.get_bikes()))
print("Total stock value:", shop.totalCost())
```
